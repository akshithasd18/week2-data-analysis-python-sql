# Week 2 – Python for Data Analysis

Dataset: 200-row sales dataset supplied for the SQL assignment. The CSV file is in this folder.

```python
import pandas as pd

df = pd.read_csv('sales_data.csv')
df.head()
```

## 1. Load a CSV file using Pandas and display basic info

```python
print('Dataset shape:', df.shape)

print('\nColumn names:')
print(df.columns.tolist())

print('\nData types and non-null values:')
df.info()

print('\nSummary statistics:')
print(df.describe(include='all'))
```

## 2. Handle missing values and duplicates using Pandas

```python
print('Missing values before cleaning:')
print(df.isna().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numeric values with the column median
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with the most frequent value
for col in df.select_dtypes(include='object').columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print('\nMissing values after cleaning:')
print(df.isna().sum())
print('Duplicate rows remaining:', df.duplicated().sum())
```

## 3. Group data by category and find total revenue

```python
category_revenue = df.groupby('category', as_index=False)['total_price'].sum()
category_revenue = category_revenue.sort_values('total_price', ascending=False)
print(category_revenue)
```

## 4. Sort data by multiple columns using Python

```python
sorted_sales = df.sort_values(
    by=['region', 'total_price'],
    ascending=[True, False]
)
print(sorted_sales)
```

## 5. Create a correlation matrix for numerical columns

```python
correlation_matrix = df.select_dtypes(include='number').corr()
print(correlation_matrix)
```

## Save cleaned dataset (optional)

```python
df.to_csv('cleaned_sales_data.csv', index=False)
print('Cleaned file saved.')
```
